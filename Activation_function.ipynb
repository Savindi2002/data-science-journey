```python
import matplotlib.pyplot as plt
import numpy as np

class Perceptron:
  """Single-Layer Perceptron classifier using the Rosenblatt learning rule."""

  def __init__(self, learning_rate: float = 0.1, epochs: int = 20):
    self.lr = learning_rate
    self.epochs = epochs
    self.weights = None
    self.bias = None
    self.error_per_epoch = []

  def fit(self, X: np.ndarray, y: np.ndarray) -> 'Perceptron':
    """Train perceptron on input matrix x and target array y."""
    n_samples, n_features = X.shape
    self.weights = np.zeros(n_features)
    self.bias = 0.0
    self.error_per_epoch = []

    for epoch in range(self.epochs):
      total_errors = 0
      for i in range(n_samples):
        linear_output = np.dot(X[i], self.weights) + self.bias
        y_pred = 1 if linear_output >= 0 else 0

        update = self.lr * (y[i] - y_pred)
        if update != 0.0:
          self.weights += update * X[i]
          self.bias += update
          total_errors += 1
      self.error_per_epoch.append(total_errors)
    return self

  def predict(self, X: np.ndarray) -> np.ndarray:
    """Predict binary class labels for sample in X."""
    linear_output = np.dot(X, self.weights) + self.bias
    return np.where(linear_output >= 0, 1, 0)

X_and = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])

model = Perceptron(learning_rate=0.01, epochs=10)
model.fit(X_and, y_and)

print("===AND Gate Results===")
print(" Learned Weights:", model.weights)
print(" Learned_Bias:", model.bias)
print("Predictions:", model.predict(X_and))
print("Error per epoch:", model.error_per_epoch)
```
