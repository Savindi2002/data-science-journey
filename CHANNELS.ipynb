```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data={
    'drink':['Latte','Espresso','Tea'],
    'total_sales':[500,200,300],
    'wait_time_mins':[5,2,3]
}
```

```python
df=pd.DataFrame(data)
```

```python
fig,axes=plt.subplots(1,2,figsize=(14,5))
sns.scatterplot(ax=axes[0],
                data=df,
                x='drink',y='wait_time_mins',
                hue='total_sales',
                palette='Blues',
                s=400)
axes[0].set_title('BAD Chart')
axes[0].set_ylabel('wait_Time')
axes[0].set_xlabel('Drink')


sns.barplot(ax=axes[1],
                data=df,
                x='drink',y='wait_time_mins',
                hue='total_sales',
                palette='crest',
                legend=False)
axes[1].set_title('good Chart')
axes[1].set_ylabel('wait_Time')
axes[1].set_xlabel('Drink')

plt.tight_layout()
plt.show()


```

```python
df_diamonds=sns.load_dataset("diamonds")
fig,ax=plt.subplots(figsize=(14,5))

sns.boxplot(ax=ax,
            data=df_diamonds,
            x="cut",
            y="price",
           palette="Set2"



)

plt.show()
```

```python

avg_price = df_diamonds.groupby("cut")["price"].mean()

plt.figure(figsize=(8, 8))
plt.pie(
    avg_price,
    labels=avg_price.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Average Diamond Price ")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

avg_price_df = avg_price.reset_index()
colors = ["#FF6B6B", "#4ECDC4", "#CC5500", "#FFBF00", "#98D8AA"]

sns.barplot(
    data=avg_price_df,
    x="cut",
    y="price",
    palette=colors
)

plt.title("Average Diamond Price")
plt.xlabel("Cut ")
plt.ylabel("Average Price ")

plt.show()
```
